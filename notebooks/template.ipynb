# Remember 실행 템플릿

저장소 루트 또는 notebooks 폴더에서 실행합니다. 공통 카탈로그의 과제·크기·모델·예산·환경을 선택합니다.
기본은 GRT 인코더의 CPU 2-update 실행 예시입니다. 디코더는 attention_mode와 objective의 두 `--set` 옵션을 제거하고, RMT는 `--model rmt`로 선택합니다.
`segment_len`(기본4)과 `num_registers`(기본32)는 RMT/GRT 공통으로 적용하는 양의 정수입니다. Remember는 segment_len=key_size+value_size+2 계약을 따르므로 길이 변경 시 데이터의 key/value 길이도 맞춰야 합니다. 조건을 바꾸면 새 output 경로를 사용합니다.
WikiText/T4 × 2는 `kaggle.ipynb`를 사용합니다. 확장 규칙은 `docs/20261011_experiment_workflow.md`를 따릅니다.


```python
from pathlib import Path
import json, subprocess, sys

repo = Path.cwd()
if not (repo / 'configs/experiments.yaml').exists():
    repo = repo.parent
output = repo / 'runs/remember_encoder_cpu'
segment_len = 4
num_registers = 32
command = [sys.executable, 'scripts/train.py', '--task', 'remember',
           '--model', 'grt', '--size', 'tiny', '--budget', 'smoke', '--runtime', 'cpu',
           '--steps', '2', '--set', 'model.attention_mode=encoder',
           '--set', 'data.options.objective=parallel',
           '--set', f'model.segment_len={segment_len}',
           '--set', f'model.num_registers={num_registers}',
           '--output-dir', str(output), '--device', 'cpu']
```

```python
subprocess.run(command, cwd=repo, check=True)
```

```python
print(json.loads((output / 'summary.json').read_text()))
```
